````python
import os

def get_api_key_from_user(env_var_name, prompt_message):
    """Get API key from environment or prompt user if not set."""
    api_key = os.getenv(env_var_name)
    if not api_key:
        api_key = input(prompt_message).strip()
        os.environ[env_var_name] = api_key  # Store it in environment for later use
    return api_key

# Retrieve OpenAI API Key
openai_api_key = get_api_key_from_user("OPENAI_API_KEY", "Please enter your OpenAI API Key: ")

# Retrieve Serper API Key
serper_api_key = get_api_key_from_user("SERPER_API_KEY", "Please enter your Serper API Key: ")

# Set OpenAI Model Name
os.environ["OPENAI_MODEL_NAME"] = 'gpt-4-turbo'

# Print confirmation (optional)
print("OpenAI API Key and Serper API Key have been set successfully.")
````

## crewAI Tools

````python
from crewai_tools import (
  FileReadTool,
  ScrapeWebsiteTool,
  MDXSearchTool,
  SerperDevTool,
  CodeDocsSearchTool,
)

search_tool = SerperDevTool()
scrape_tool = ScrapeWebsiteTool()
read_resume = FileReadTool(file_path='./fake_resume.md')
semantic_search_resume = MDXSearchTool(mdx='./fake_resume.md')
code_tool = CodeDocsSearchTool()
````

## Creating Agents

````python
Architect = Agent(
    role="Senior Architect",
    goal="Design the web application which should be implemented by the whole team {app} "
         "Develop the backend, database and security access management for the different users /admins"
         "Develop the mobile phone clients / running on android.",
    tools = [scrape_tool, search_tool, code_tool],
    verbose=True,
   backstory = (
        "You are a Senior Architect responsible for designing and overseeing the development of web applications. "
        "Your role encompasses creating the application's architecture, including backend systems, databases, "
        "and security access management for various user roles (users/admins). "
        "Additionally, you guide the team in implementing mobile clients running on Android platforms. "
        "Core Skills: Proficiency in system design, microservices architecture, and distributed systems. "
        "Backend Expertise: Advanced knowledge of Node.js, Python, and Java for scalable backend systems. "
        "Databases: Expert in relational databases like PostgreSQL and MySQL, as well as NoSQL databases like MongoDB. "
        "Security: Strong understanding of OAuth, role-based access control (RBAC), and secure API design. "
        "Mobile Development: Proficient in guiding the development of Android applications using Kotlin or Flutter. "
        "DevOps: Skilled in Kubernetes, Docker, and CI/CD pipelines for deploying and maintaining systems. "
        "Qualifications and Skills Sought: Proven leadership in managing cross-functional teams, "
        "expertise in scalable system design, and the ability to mentor developers across frontend, backend, and mobile disciplines."
    ),
    allow_code_execution = True,
    allow_delegation=True

)
````

````python
# Agent 1: Senior Front End
FrontEnder1 = Agent(
    role="Senior Frontend",
    goal="Write a frontend code for web application for the given {app} "
         "Check which is the most suitable frontend tool for you",
    tools = [scrape_tool, search_tool, code_tool],
    verbose=True,
    backstory=(
        "You are an expert frontend who can implement frontend code for webapplications "
        "Core Skills: Semantic HTML, CSS, JavaScript (ES6+), responsive design, and accessibility (WCAG compliance)."
        "Frameworks & Libraries: Proficient in React.js, Next.js, AngularJS, and jQuery (legacy systems)."
        "Languages: Advanced knowledge of TypeScript for scalable and type-safe applications. "
        "qualifications and skills sought "
        "UI Frameworks: Experience with Bootstrap, Material-UI, and Chakra UI"
        "effective application tailoring."
    ),
    allow_code_execution = True,
    allow_delegation=True

)
````

````python
# Agent 2: Junior Front end
FrontEnder2 = Agent(
    role="Junior Frontend",
    goal="Write a frontend code for web application for the given {app} "
         "Check which is the most suitable frontend tool for you",
    tools = [scrape_tool, search_tool, code_tool],
    verbose=True,
    backstory=(
        "You are a Junior frontend who can implement frontend code for webapplications "
        "Core Skills: HTML5, CSS3, JavaScript (ES6+), and basic TypeScript."
        "Frameworks & Libraries: React.js (entry-level), working knowledge of Next.js for routing and server-side rendering."
        "Languages: Advanced knowledge of TypeScript for scalable and type-safe applications. "
        "qualifications and skills sought "
        "UI Frameworks: Experience with Bootstrap, Material-UI, and Chakra UI"
        "Styling: Proficient in using CSS frameworks like Bootstrap and Tailwind CSS for rapid UI development."
    ),
    allow_code_execution = True,
    allow_delegation=False
)
````

````python
# Agent 3: Backend senior
Backender1 = Agent(
    role="Senior Backend",
    goal="Write a Backend code for web application for the given {app} "
         "Check which is the most suitable Backend tool for you",
    tools = [scrape_tool, search_tool, code_tool],
    verbose=True,
    backstory = (
        "You are a Senior Backend Engineer responsible for designing scalable, secure, and performant backend architectures for web applications. "
        "Core Skills: Advanced knowledge of Node.js, Python, and Go for backend development. "
        "Frameworks: Expertise in Django, FastAPI for Python, and Nest.js for complex Node.js applications. "
        "Databases: Proficient in SQL databases like PostgreSQL and MySQL, as well as NoSQL solutions like MongoDB and Redis. "
        "DevOps Tools: Familiarity with Kubernetes, Docker for container orchestration, and CI/CD tools like Jenkins and GitHub Actions. "
        "Security: Strong skills in implementing authentication, authorization, and data encryption. "
        "Qualifications and Skills Sought: Strategic thinker with a focus on scalability, mentor to junior developers, "
        "and experienced in microservices architecture and system design."
    ),
    allow_code_execution = True,
    allow_delegation=True
)
````

````python
# Agent 4: Backend Junior
Backender2 = Agent(
    role="Junior Backend",
    goal="Write a Backend code for web application for the given {app} "
         "Check which is the most suitable Backend tool for you",
    tools = [scrape_tool, search_tool, code_tool],
    verbose=True,
    backstory = (
        "You are a Junior Backend Engineer responsible for developing and maintaining backend code for web applications. "
        "Core Skills: Proficient in Python and Node.js for backend development. "
        "Frameworks: Express.js for RESTful APIs and Flask for lightweight backend projects. "
        "Databases: Experience with MongoDB and PostgreSQL for data storage and retrieval. "
        "Tools: Familiarity with Postman for API testing, Docker for containerization, and Git for version control. "
        "Other Skills: Basic understanding of authentication mechanisms like JWT and CI/CD pipelines. "
        "Qualifications and Skills Sought: Strong debugging abilities, eagerness to learn advanced backend patterns, "
        "and the ability to collaborate effectively with the frontend team."
    ),
    allow_code_execution = True,
    allow_delegation=True
)
````

````python
# Agent 2: Profiler
profiler = Agent(
    role="Personal Profiler for Engineers",
    goal="Do increditble research on job applicants "
         "to help them stand out in the job market",
    tools = [scrape_tool, search_tool,
             read_resume, semantic_search_resume],
    verbose=True,
    backstory=(
        "Equipped with analytical prowess, you dissect "
        "and synthesize information "
        "from diverse sources to craft comprehensive "
        "personal and professional profiles, laying the "
        "groundwork for personalized resume enhancements."
    )
)
````

````python
# Agent 3: Resume Strategist
resume_strategist = Agent(
    role="Resume Strategist for Engineers",
    goal="Find all the best ways to make a "
         "resume stand out in the job market.",
    tools = [scrape_tool, search_tool,
             read_resume, semantic_search_resume],
    verbose=True,
    backstory=(
        "With a strategic mind and an eye for detail, you "
        "excel at refining resumes to highlight the most "
        "relevant skills and experiences, ensuring they "
        "resonate perfectly with the job's requirements."
    )
)
````

````python
# Agent 4: Interview Preparer
interview_preparer = Agent(
    role="Engineering Interview Preparer",
    goal="Create interview questions and talking points "
         "based on the resume and job requirements",
    tools = [scrape_tool, search_tool,
             read_resume, semantic_search_resume],
    verbose=True,
    backstory=(
        "Your role is crucial in anticipating the dynamics of "
        "interviews. With your ability to formulate key questions "
        "and talking points, you prepare candidates for success, "
        "ensuring they can confidently address all aspects of the "
        "job they are applying for."
    )
)
````

## Creating Tasks

````python
# Task for Researcher Agent: Extract Job Requirements
arch_task = Task(
    description=(
        "Design the architecture for a web application that ensures scalability, security, and maintainability. "
        "Define the backend structure, database schemas, and security access management for users and admins. "
        "Collaborate with the frontend, backend, and mobile development teams to align on the implementation strategy. "
        "Guide the development of Android mobile clients, ensuring smooth integration with backend APIs. "
        "Establish DevOps workflows for deployment, monitoring, and scaling the application."
    ),
    expected_output=(
        "A detailed architectural design document outlining the system's structure, backend and database specifications, "
        "security protocols, and integration points for the frontend and mobile applications. "
        "A fully functional web application with role-based access control and an Android mobile client. "
        "CI/CD pipelines for efficient deployment and scaling."
    ),
    agent=FrontEnder1,
    async_execution=True
)
````

````python
# Task for Researcher Agent: Extract Job Requirements
Front_end_sen = Task(
    description=(
        "Develop a scalable and reusable component library for the web application, "
        "ensuring alignment with the existing design system. "
        "Collaborate with the design team to translate UI mockups into pixel-perfect, "
        "responsive components. Optimize the components for performance, "
        "cross-browser compatibility, and accessibility (WCAG compliance)."
    ),
    expected_output=(
        "A comprehensive library of reusable UI components implemented in React.js "
        "or Next.js, styled with Tailwind CSS or Material-UI. "
        "Each component should include unit tests, documentation, and "
        "examples for integration into the application."
    ),
    agent=FrontEnder1,
    async_execution=True,
    context=[arch_task]
)
````

````python
# Task for Researcher Agent: Extract Job Requirements
Front_end_jun = Task(
    description=(
        "Implement responsive UI components based on provided design mockups. "
        "Ensure proper integration of API endpoints into the frontend, "
        "and handle data rendering and state management with React.js. "
        "Collaborate with the senior frontend engineer to address feedback "
        "and optimize components for usability and performance."
    ),
    expected_output=(
        "Functional and responsive UI components that match the provided designs. "
        "Properly integrated API calls to fetch and display data dynamically. "
        "Code adhering to best practices, including comments and basic unit tests."
    ),
    agent=FrontEnder2,
    async_execution=False,
    context=[arch_task, Front_end_sen]
)
````

````python
# Task for Researcher Agent: Extract Job Requirements
back_end_sen = Task(
    description=(
        "Design and implement a scalable backend architecture to handle high traffic for the web application. "
        "Develop GraphQL and RESTful APIs, ensuring flexibility for frontend integration. "
        "Establish database schemas and optimize data pipelines for performance. "
        "Mentor the junior backend engineer and enforce coding standards across the backend team."
    ),
    expected_output=(
        "A scalable and well-documented backend architecture supporting the web application. "
        "GraphQL and RESTful APIs implemented with secure authentication and authorization. "
        "Optimized database schemas and data pipelines, with clear guidance and mentorship provided to junior team members."
    ),
    agent=Backender1,
    async_execution=True,
    context=[arch_task]
)
````

````python
# Task for Researcher Agent: Extract Job Requirements
back_end_jun = Task(
    description=(
        "Develop and maintain RESTful API endpoints to support frontend functionalities. "
        "Write and optimize database queries for efficient data retrieval and storage. "
        "Ensure the APIs are well-documented, tested, and secure. "
        "Collaborate with the senior backend engineer to implement robust backend solutions."
    ),
    expected_output=(
        "Fully functional API endpoints with secure authentication mechanisms. "
        "Optimized database queries for handling data efficiently. "
        "Comprehensive API documentation and automated tests for all endpoints."
    ),
    agent=Backender2,
    async_execution=False,
    Context = [Front_end_sen, back_end_sen]
)
````

````python
# Task for Profiler Agent: Compile Comprehensive Profile
profile_task = Task(
    description=(
        "Compile a detailed personal and professional profile "
        "using the GitHub ({github_url}) URLs, and personal write-up "
        "({personal_writeup}). Utilize tools to extract and "
        "synthesize information from these sources."
    ),
    expected_output=(
        "A comprehensive profile document that includes skills, "
        "project experiences, contributions, interests, and "
        "communication style."
    ),
    agent=profiler,
    async_execution=True
)
````

- You can pass a list of tasks as `context` to a task.
- The task then takes into account the output of those tasks in its execution.
- The task will not run until it has the output(s) from those tasks.

````python
# Task for Resume Strategist Agent: Align Resume with Job Requirements
resume_strategy_task = Task(
    description=(
        "Using the profile and job requirements obtained from "
        "previous tasks, tailor the resume to highlight the most "
        "relevant areas. Employ tools to adjust and enhance the "
        "resume content. Make sure this is the best resume even but "
        "don't make up any information. Update every section, "
        "inlcuding the initial summary, work experience, skills, "
        "and education. All to better reflrect the candidates "
        "abilities and how it matches the job posting."
    ),
    expected_output=(
        "An updated resume that effectively highlights the candidate's "
        "qualifications and experiences relevant to the job."
    ),
    output_file="tailored_resume.md",
    context=[research_task, profile_task],
    agent=resume_strategist
)
````

````python
# Task for Interview Preparer Agent: Develop Interview Materials
interview_preparation_task = Task(
    description=(
        "Create a set of potential interview questions and talking "
        "points based on the tailored resume and job requirements. "
        "Utilize tools to generate relevant questions and discussion "
        "points. Make sure to use these question and talking points to "
        "help the candiadte highlight the main points of the resume "
        "and how it matches the job posting."
    ),
    expected_output=(
        "A document containing key questions and talking points "
        "that the candidate should prepare for the initial interview."
    ),
    output_file="interview_materials.md",
    context=[research_task, profile_task, resume_strategy_task],
    agent=interview_preparer
)
````

## Creating the Crew

````python
web_application_crew = Crew(
    agents=[Architect,
            FrontEnder1,
            FrontEnder2,
            Backender1,
            Backender2
           ],

    tasks=[arch_task,
           Front_end_sen,
           Front_end_jun,
           back_end_sen,
           back_end_jun
           ],

    verbose=True
)
````

## Running the Crew

- Set the inputs for the execution of the crew.

````python
web_application_inputs = {

    'Create a full application, provide the full frontend and backend code according to the following: project_overview': (
        "The team will collaborate to design and develop a fully functional web application. "
        "The project will include a scalable backend, a responsive and user-friendly frontend, "
        "and an Android mobile client. The Senior Architect will lead the architecture design, "
        "while the Senior and Junior Developers will implement the backend and frontend components."
    ),
    'team_roles': {
        'Architect': (
            "Design the entire architecture of the web application, including backend, "
            "database schemas, API structures, and security protocols. Define the frontend-backend integration "
            "and oversee the alignment of the mobile client with the backend. Collaborate with all developers "
            "to ensure smooth implementation of the design."
        ),
        'Senior Frontend Developer': (
            "Develop a reusable component library for the frontend, ensuring scalability and "
            "alignment with the design system. Optimize the application for performance, cross-browser compatibility, "
            "and accessibility. Collaborate with the backend team to integrate APIs seamlessly."
        ),
        'Junior Frontend Developer': (
            "Implement responsive UI components from the mockups provided by the design team. "
            "Work on data rendering and state management using React.js, ensuring pixel-perfect designs. "
            "Collaborate with the Senior Frontend Developer for performance optimizations and API integration."
        ),
        'Senior Backend Engineer': (
            "Design and implement the backend infrastructure, including RESTful APIs and database schemas. "
            "Ensure scalability, security, and optimal performance for the application. Mentor the Junior Backend Engineer "
            "and collaborate with the frontend team for seamless integration."
        ),
        'Junior Backend Engineer': (
            "Develop and maintain APIs as per the specifications provided by the Senior Backend Engineer. "
            "Handle database interactions, ensuring efficient data storage and retrieval. Test and debug backend functionalities "
            "and document the APIs for the frontend developers."
        ),
    },
    'project_tools': {
        'Frontend': ["React.js", "Next.js", "Tailwind CSS", "Material-UI"],
        'Backend': ["Node.js", "Express.js", "Django", "PostgreSQL", "MongoDB"],
        'DevOps': ["Docker", "Kubernetes", "GitHub Actions", "AWS"],
        'Mobile': ["Kotlin", "Flutter"],
    },
    'expected_output': (
        "A fully functional web application with: "
        "1. A scalable backend supporting user authentication, database interactions, and API endpoints. "
        "2. A responsive and accessible frontend implementing reusable components and state management. "
        "3. An Android mobile client integrated with backend APIs for seamless user experiences. "
        "4. Comprehensive documentation for all components and CI/CD pipelines for automated deployment."
    ),
}
````

**Note**: LLMs can provide different outputs for they same input, so what you get might be different than what you see in the video.

````python
### this execution will take a few minutes to run
result = web_application_crew.kickoff(inputs={**web_application_inputs, "app": "Time planner"})
````
